```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import to_categorical
import time

# -----------------------------------------------------------
# 1. Mengambil dataset MNIST dan melakukan preprocessing awal
# -----------------------------------------------------------
(trainX, trainY), (testX, testY) = mnist.load_data()

# Normalisasi piksel gambar ke rentang 0–1
trainX = trainX.astype("float32") / 255.0
testX  = testX.astype("float32")  / 255.0

# Mengubah label menjadi format one-hot
trainY = to_categorical(trainY, 10)
testY  = to_categorical(testY, 10)

print(f"Dimensi data train  : {trainX.shape}")
print(f"Dimensi data testing: {testX.shape}")
print("-" * 50)


# -----------------------------------------------------------
# Fungsi umum untuk membuat, melatih, dan mengevaluasi model
# -----------------------------------------------------------
def build_and_test(model, label):
    model.compile(
        optimizer="adam",
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    start = time.time()
    history = model.fit(
        trainX, trainY,
        epochs=10,
        batch_size=128,
        validation_split=0.1,
        verbose=0  # supaya output training tidak memenuhi terminal
    )
    duration = time.time() - start

    loss, accuracy = model.evaluate(testX, testY, verbose=0)

    print(label)
    print(f"   Akurasi      : {accuracy:.4f}")
    print(f"   Durasi latih : {duration:.2f} detik")
    print(f"   Loss testing : {loss:.4f}")
    print("-" * 50)

    return accuracy, duration


# -----------------------------------------------------------
# Kumpulan percobaan arsitektur model
# -----------------------------------------------------------

# Percobaan 1: Model dasar (2 hidden layer dengan ReLU)
modelA = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation="relu"),
    Dense(64, activation="relu"),
    Dense(10, activation="softmax")
])
accA, timeA = build_and_test(modelA, "1. Model Dasar (128 → 64, ReLU)")

# Percobaan 2: Hidden layer lebih besar (256 → 128)
modelB = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(256, activation="relu"),
    Dense(128, activation="relu"),
    Dense(10, activation="softmax")
])
accB, timeB = build_and_test(modelB, "2. Hidden Besar (256 → 128, ReLU)")

# Percobaan 3: Menambah hidden layer baru (3 hidden layer)
modelC = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(256, activation="relu"),
    Dense(128, activation="relu"),
    Dense(64, activation="relu"),  # tambahan layer
    Dense(10, activation="softmax")
])
accC, timeC = build_and_test(modelC, "3. 3 Hidden Layer (256 → 128 → 64, ReLU)")

# Percobaan 4: Menggunakan aktivasi Sigmoid
modelD = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation="sigmoid"),
    Dense(64, activation="sigmoid"),
    Dense(10, activation="softmax")
])
accD, timeD = build_and_test(modelD, "4. Sigmoid (128 → 64)")

# Percobaan 5: Sigmoid dengan neuron lebih banyak
modelE = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(256, activation="sigmoid"),
    Dense(128, activation="sigmoid"),
    Dense(10, activation="softmax")
])
accE, timeE = build_and_test(modelE, "5. Sigmoid Besar (256 → 128)")

# -----------------------------------------------------------
# Ringkasan dari seluruh percobaan
# -----------------------------------------------------------
print("\n" + "=" * 60)
print("RINGKASAN HASIL EKSPERIMEN")
print("=" * 60)
print(f"{'No':<3} {'Konfigurasi':<35} {'Akurasi':<10} {'Waktu (detik)':<15}")
print("-" * 60)
print(f"1   {'Dasar (128→64 ReLU)':<35} {accA:.4f}     {timeA:6.2f}")
print(f"2   {'256→128 ReLU':<35} {accB:.4f}     {timeB:6.2f}")
print(f"3   {'3 layer ReLU':<35} {accC:.4f}     {timeC:6.2f}")
print(f"4   {'Sigmoid 128→64':<35} {accD:.4f}     {timeD:6.2f}")
print(f"5   {'Sigmoid 256→128':<35} {accE:.4f}     {timeE:6.2f}")
print("=" * 60)
```
